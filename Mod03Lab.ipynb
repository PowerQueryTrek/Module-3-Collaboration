```python
class Vehicle:
    def __init__(self, vehicle_type):
        self.vehicle_type = vehicle_type

class Automobile(Vehicle):
    def __init__(self, year, make, model, doors, roof):
        super().__init__("car")
        self.year = year
        self.make = make
        self.model = model
        self.doors = doors
        self.roof = roof

print("Enter car details")
year = input("Year: ")
make = input("Make: ")
model = input("Model: ")
doors = input("Number of doors (2 or 4): ")
roof = input("Type of roof (solid or sun roof): ")

a = Automobile(year, make, model, doors, roof)

print()
print("Vehicle type:", a.vehicle_type)
print("Year:", a.year)
print("Make:", a.make)
print("Model:", a.model)
print("Number of doors:", a.doors)
print("Type of roof:", a.roof)
```
