Sort 0s, 1s and 2s

```python
def sort012(arr):
    zeros = []
    ones = []
    twos = []
    for x in arr:
        if x == 0:
            zeros.append(x)
        elif x == 1:
            ones.append(x)
        else:
            twos.append(x)
    return zeros + ones + twos
    
arr = [0, 1, 1, 0, 1, 2, 1, 2, 0, 0, 0, 1]
print(sort012(arr))
```

First Occurrence in Sorted

```python
class Solution:
    def firstSearch(self, arr, k):
        if arr == []:
            result = -1
        else:
            startindex = 0
            endindex = len(arr) - 1
            while startindex != endindex:
                midindex = (startindex + endindex) // 2
                if arr[midindex] < k:
                    startindex = midindex + 1
                else:
                    endindex = midindex
            if arr[startindex] == k:
                result = startindex
            else:
                result = -1
    
        return result

arr = [1,1,1,3]
k = 2
print(Solution().firstSearch(arr, k))
```
